```python
##------------------------------------------------------------------------------
## Tập tin 'President Joe Biden.txt' lưu trữ tin tức về chuyến viếng thăm VN
## của Tổng thống Hoa Kỳ Joe Biden
##------------------------------------------------------------------------------

## Luu y: doan code duoi day gia dinh tap tin 'President Joe Biden.txt' da
## duoc tai len cung thu muc voi notebook nay.
TEN_TAP_TIN = "President Joe Biden.txt"
```

```python
## Khong su dung cac ham tinh kich thuoc cua tap tin, hay cho biet tap tin
##     co chua bao nhieu ky tu.
##
## Viet code:
so_ky_tu = 0
with open(TEN_TAP_TIN, encoding="utf-8") as f:
    for dong in f:
        for _ in dong:
            so_ky_tu += 1

print(f"Tap tin co {so_ky_tu} ky tu")
```

```python
## Hay cho biet dong dai nhat trong tap tin.
##
## Viet code:
with open(TEN_TAP_TIN, encoding="utf-8") as f:
    cac_dong = f.readlines()

dong_dai_nhat = max(cac_dong, key=len)
print(f"Dong dai nhat ({len(dong_dai_nhat.rstrip())} ky tu): {dong_dai_nhat.strip()}")
```

```python
## Liet ke danh sach cac tu co trong tap tin (sap xep tang dan).
##
## Viet code:
import re

with open(TEN_TAP_TIN, encoding="utf-8") as f:
    noi_dung = f.read()

danh_sach_tu = re.findall(r"[A-Za-zÀ-ỹ]+", noi_dung)
danh_sach_tu_sap_xep = sorted(set(danh_sach_tu))
print(danh_sach_tu_sap_xep)
```

```python
## Liet ke cac tu co it nhat 2 ky tu giong nhau (khong phan biet chu hoa, chu thuong).
##     VD: increasingly, President, ...
##     Sap xep ket qua tang dan.
##
## Viet code:
def co_it_nhat_2_ky_tu_giong_nhau(tu):
    tu_thuong = tu.lower()
    dem = {}
    for ky_tu in tu_thuong:
        dem[ky_tu] = dem.get(ky_tu, 0) + 1
        if dem[ky_tu] >= 2:
            return True
    return False

ket_qua = sorted(set(tu for tu in danh_sach_tu if co_it_nhat_2_ky_tu_giong_nhau(tu)))
print(ket_qua)
```

```python
## Hay cho biet tu nao co chua ky tu 'e' xuat hien nhieu nhat.
##
## Viet code:
tu_nhieu_e_nhat = max(danh_sach_tu, key=lambda t: t.lower().count('e'))
print(f"Tu chua nhieu 'e' nhat: {tu_nhieu_e_nhat} ({tu_nhieu_e_nhat.lower().count('e')} lan)")


## Hien thi so thu tu va noi dung cua nhung dong chua tu tim duoc trong cau tren
##
## Viet code:
for i, dong in enumerate(cac_dong, start=1):
    if tu_nhieu_e_nhat.lower() in dong.lower():
        print(f"Dong {i}: {dong.strip()}")
```

```python
## Hay cho biet tu nao xuat hien nhieu nhat trong tap tin.
##
## Viet code:
from collections import Counter

dem_tu = Counter(tu.lower() for tu in danh_sach_tu)
tu_xuat_hien_nhieu_nhat, so_lan = dem_tu.most_common(1)[0]
print(f"Tu xuat hien nhieu nhat: '{tu_xuat_hien_nhieu_nhat}' ({so_lan} lan)")


## Hien thi so thu tu va noi dung cua nhung dong chua tu tim duoc trong cau tren
##
## Viet code:
for i, dong in enumerate(cac_dong, start=1):
    if tu_xuat_hien_nhieu_nhat in dong.lower():
        print(f"Dong {i}: {dong.strip()}")
```
