##### *Cấu trúc rẽ nhánh IF*

```python
##------------------------------------------------------------------------------
## Nhập điểm của học sinh và hiển thị xếp loại Giỏi, Khá, ...
##------------------------------------------------------------------------------
## Viết code:
def xep_loai(diem):
  if diem >= 8.5:
    return "A"
  elif diem >= 7.0 and diem < 8.5:
    return "B"
  elif diem >= 5.0 and diem < 7.0:
    return "C"
  elif diem < 5.0 and diem >= 3.0:
    return "D"
  else:
    return "F"

def nhap_diem():
  while True:
    try:
      diem = float(input("Nhập điểm của bạn (từ 0 đến 10): "))
      if 1 <= diem <= 10:
        return diem
      else:
        print("Vui lòng nhập đúng số từ 0 đến 10!")
    except ValueError:
      print("Gía trị không hợp lệ. Vui lòng nhập lại số! ")

diem = nhap_diem()
loai = xep_loai(diem)
print(f"Điểm của bạn: {diem}: xếp loại {loai}")
```

```python
##------------------------------------------------------------------------------
## Nhập 2 số và hiển thị kết quả so sánh
##------------------------------------------------------------------------------
## Viết code:
def nhap_so():
  while True:
    try:
      so1 = float(input("Nhập số 1: "))
      so2 = float(input("Nhập số 2: "))
      return so1, so2
    except ValueError:
      print("Gía trị không đúng. Vui lòng nhập 1 con số!")

def so_sanh(so1, so2):
  if so1 < so2:
    return f"Số {so1} NHỎ hơn Số {so2}"
  elif so1 > so2:
    return f"Số {so1} LỚN hơn Số {so2}"
  else:
    return f"Số {so1} BẰNG Số {so2}"

so1, so2 = nhap_so()
ket_qua = so_sanh(so1, so2)
print(f"Kết quả so sánh của {so1} và {so2}: {ket_qua}")
```

**2. Dien tich da giac loi khi biet n dinh (cong thuc Shoelace)**

```python
def dien_tich_da_giac(dinh):
    n = len(dinh)
    S = 0
    for i in range(n):
        x1, y1 = dinh[i]
        x2, y2 = dinh[(i + 1) % n]
        S += x1 * y2 - x2 * y1
    return abs(S) / 2

# Vi du: hinh vuong don vi
dinh = [(0, 0), (4, 0), (4, 4), (0, 4)]
print(f"Dien tich = {dien_tich_da_giac(dinh)}")
```

**3. Giai phuong trinh bac 2: ax^2 + bx + c = 0**

```python
import math

def giai_pt_bac2(a, b, c):
    if a == 0:
        if b == 0:
            return "Vo nghiem" if c != 0 else "Vo so nghiem"
        return [-c / b]
    delta = b ** 2 - 4 * a * c
    if delta < 0:
        return "Vo nghiem thuc"
    elif delta == 0:
        return [-b / (2 * a)]
    else:
        sq = math.sqrt(delta)
        return [(-b + sq) / (2 * a), (-b - sq) / (2 * a)]

a = float(input("Nhap a: "))
b = float(input("Nhap b: "))
c = float(input("Nhap c: "))
print("Nghiem:", giai_pt_bac2(a, b, c))
```

**1. To hop C(n, k) = n(n-1)...(n-k+1) / k!**

```python
def to_hop(n, k):
    if k == 0:
        return 1
    return to_hop(n, k - 1) * (n - k + 1) // k

print(to_hop(5, 2))
```

**2. So Fibonacci thu n**

```python
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

print([fibonacci(i) for i in range(10)])
```

**3. BCNN cua 2 so nguyen (de quy)**

```python
def uscln(a, b):
    if b == 0:
        return a
    return uscln(b, a % b)

def bcnn(a, b):
    return abs(a * b) // uscln(a, b)

print(bcnn(4, 6))
```

**4. Tong, max, min cua day a0,...,an-1 (de quy)**

```python
def tong_de_quy(a, n):
    if n == 0:
        return 0
    return a[n - 1] + tong_de_quy(a, n - 1)

def max_de_quy(a, n):
    if n == 1:
        return a[0]
    return max(a[n - 1], max_de_quy(a, n - 1))

def min_de_quy(a, n):
    if n == 1:
        return a[0]
    return min(a[n - 1], min_de_quy(a, n - 1))

a = [3, 7, 1, 9, 4]
print(tong_de_quy(a, len(a)), max_de_quy(a, len(a)), min_de_quy(a, len(a)))
```

**5. Kiem tra day a0,...,an-1 co phai day tang (de quy)**

```python
def la_day_tang(a, n):
    if n <= 1:
        return True
    return a[n - 2] <= a[n - 1] and la_day_tang(a, n - 1)

a = [1, 2, 2, 5, 9]
print(la_day_tang(a, len(a)))
```

**6. Vi tri gia tri x trong day a0,...,an-1 (de quy), -1 neu khong co**

```python
def tim_vi_tri(a, n, x):
    if n == 0:
        return -1
    if a[n - 1] == x:
        return n - 1
    return tim_vi_tri(a, n - 1, x)

a = [5, 3, 8, 1, 9]
print(tim_vi_tri(a, len(a), 8))
```

**7. Nhu cau 6 nhung day co thu tu giam (tim kiem nhi phan de quy)**

```python
def tim_kiem_nhi_phan_giam(a, lo, hi, x):
    if lo > hi:
        return -1
    mid = (lo + hi) // 2
    if a[mid] == x:
        return mid
    elif a[mid] > x:
        return tim_kiem_nhi_phan_giam(a, mid + 1, hi, x)
    else:
        return tim_kiem_nhi_phan_giam(a, lo, mid - 1, x)

a = [9, 8, 5, 3, 1]  # day giam
print(tim_kiem_nhi_phan_giam(a, 0, len(a) - 1, 5))
```

**10. Liet ke cac chuoi k-phan n ky hieu**

```python
def chuoi_k_phan(n, k, prefix=""):
    if n == 0:
        print(prefix)
        return
    for chu_so in range(k):
        chuoi_k_phan(n - 1, k, prefix + str(chu_so))

chuoi_k_phan(2, 3)  # n=2 vi tri, k=3 ky hieu (0,1,2)
```

**11. In tat ca hoan vi cua day n ky hieu phan biet**

```python
def hoan_vi(chuoi, da_chon=""):
    if not chuoi:
        print(da_chon)
        return
    for i in range(len(chuoi)):
        hoan_vi(chuoi[:i] + chuoi[i+1:], da_chon + chuoi[i])

hoan_vi("abc")
```

**12. In ra cac chuoi con k ky hieu cua chuoi n > k ky hieu (to hop khong lap)**

```python
def to_hop_chuoi(chuoi, k, bat_dau=0, da_chon=""):
    if k == 0:
        print(da_chon)
        return
    for i in range(bat_dau, len(chuoi)):
        to_hop_chuoi(chuoi, k - 1, i + 1, da_chon + chuoi[i])

to_hop_chuoi("1234", 2)
```

**13. In vi tri cac phan tu co tong bang T (de quy - liet ke tap con)**

```python
def tim_tong_T(a, n, T, vi_tri=None):
    if vi_tri is None:
        vi_tri = []
    if n == 0:
        if T == 0 and vi_tri:
            print(vi_tri)
        return
    # khong chon phan tu a[n-1]
    tim_tong_T(a, n - 1, T, vi_tri)
    # chon phan tu a[n-1]
    tim_tong_T(a, n - 1, T - a[n - 1], vi_tri + [n - 1])

A = [3, 4, 7, 2, 1]
tim_tong_T(A, len(A), 7)
```

**14. Doi tien menh gia lon thanh cac to tien menh gia nho**

```python
def doi_tien(so_tien, menh_gia):
    if not menh_gia:
        return {}
    mg = menh_gia[0]
    so_to = so_tien // mg
    con_lai = so_tien % mg
    kq = doi_tien(con_lai, menh_gia[1:])
    if so_to > 0:
        kq[mg] = so_to
    return kq

menh_gia = [500000, 200000, 100000, 50000, 20000, 10000]
print(doi_tien(870000, menh_gia))
```

**15. Dien tich n-giac loi tu toa do n dinh (Shoelace, de quy)**

```python
def dien_tich_da_giac_de_quy(P, i=0):
    n = len(P)
    if i == n:
        return 0
    x1, y1 = P[i]
    x2, y2 = P[(i + 1) % n]
    return (x1 * y2 - x2 * y1) + dien_tich_da_giac_de_quy(P, i + 1)

P = [(0, 0), (4, 0), (4, 4), (0, 4)]
print(f"Dien tich = {abs(dien_tich_da_giac_de_quy(P)) / 2}")
```
