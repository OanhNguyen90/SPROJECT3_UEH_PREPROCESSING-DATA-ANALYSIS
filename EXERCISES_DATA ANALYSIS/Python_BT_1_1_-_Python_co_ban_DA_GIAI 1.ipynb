```python
##------------------------------------------------------------------------------
## Tap tin 'LTPTDL_BT1.Users.txt' luu tru thong tin ve so thich cua nguoi su dung
## doi voi 4 the loai phim: hinh su, tinh cam, hoat hinh, tai lieu.
## Tap tin 'LTPTDL_BT1.Films.txt' luu tru muc do lien quan cua bo phim doi voi
## 4 the loai tren.
##------------------------------------------------------------------------------

## Luu y: doan code duoi day gia dinh 2 tap tin da duoc tai len cung thu muc
## voi notebook nay.
TEN_TAP_TIN_USERS = "LTPTDL_BT1.Users.txt"
TEN_TAP_TIN_FILMS = "LTPTDL_BT1.Films.txt"
```

```python
## Yeu cau 1) Doc noi dung hai tap tin du lieu nay.
## Moi dong: <ma so> <hinh su> <tinh cam> <hoat hinh> <tai lieu>
## Ket qua: dictionary {ma_so: [4 gia tri]}
##
## Viet code:
def doc_du_lieu(ten_tap_tin):
    du_lieu = {}
    with open(ten_tap_tin, encoding="utf-8") as f:
        for dong in f:
            phan = dong.split()
            if not phan:
                continue
            ma_so = phan[0]
            gia_tri = [float(x) for x in phan[1:5]]
            du_lieu[ma_so] = gia_tri
    return du_lieu

du_lieu_users = doc_du_lieu(TEN_TAP_TIN_USERS)
du_lieu_films = doc_du_lieu(TEN_TAP_TIN_FILMS)

print(f"So luong nguoi su dung: {len(du_lieu_users)}")
print(f"So luong bo phim: {len(du_lieu_films)}")
```

```python
## Ham tinh do phu hop giua 1 nguoi su dung va 1 bo phim: tich vo huong
## (tong cac tich thanh phan tuong ung giua vector so thich va vector the loai)
##
## Viet code:
def tinh_do_phu_hop(vector_user, vector_phim):
    return sum(a * b for a, b in zip(vector_user, vector_phim))
```

```python
## Yeu cau 2) Nhap ma so nguoi su dung va so nguyen n trong [1, 3]; hien thi n
## bo phim phu hop nhat voi nguoi dung. Lap lai nhieu lan, kiem tra tinh hop le
## cua ma so nguoi su dung khi nhap.
##
## Viet code:
while True:
    ma_so_nguoi_dung = input("Nhap ma so nguoi su dung (Enter de dung): ")
    if ma_so_nguoi_dung == "":
        break

    if ma_so_nguoi_dung not in du_lieu_users:
        print("Ma so nguoi su dung khong hop le!\n")
        continue

    try:
        n = int(input("Nhap so luong phim can goi y (1-3): "))
    except ValueError:
        print("Gia tri n khong hop le!\n")
        continue

    if n < 1 or n > 3:
        print("Gia tri n phai trong khoang [1, 3]!\n")
        continue

    vector_user = du_lieu_users[ma_so_nguoi_dung]
    danh_sach_do_phu_hop = [
        (ma_phim, tinh_do_phu_hop(vector_user, vector_phim))
        for ma_phim, vector_phim in du_lieu_films.items()
    ]
    danh_sach_do_phu_hop.sort(key=lambda x: x[1], reverse=True)
    top_n = danh_sach_do_phu_hop[:n]

    print(f"\n{n} bo phim phu hop nhat voi nguoi dung {ma_so_nguoi_dung}:")
    for ma_phim, diem in top_n:
        print(f"  Phim {ma_phim} - Diem phu hop: {diem:.2f}")
    print()
```

```python
## Yeu cau 3) Nhap ma so bo phim va hien thi tat ca nguoi su dung co the ua
## thich bo phim nay (chon nhung nguoi co diem phu hop >= diem trung binh cua
## tat ca nguoi dung doi voi bo phim do). Lap lai nhieu lan, kiem tra tinh hop
## le cua ma so bo phim khi nhap.
##
## Viet code:
while True:
    ma_so_phim = input("Nhap ma so bo phim (Enter de dung): ")
    if ma_so_phim == "":
        break

    if ma_so_phim not in du_lieu_films:
        print("Ma so bo phim khong hop le!\n")
        continue

    vector_phim = du_lieu_films[ma_so_phim]
    diem_theo_nguoi_dung = {
        ma_nguoi_dung: tinh_do_phu_hop(vector_user, vector_phim)
        for ma_nguoi_dung, vector_user in du_lieu_users.items()
    }
    diem_trung_binh = sum(diem_theo_nguoi_dung.values()) / len(diem_theo_nguoi_dung)

    nguoi_dung_ua_thich = sorted(
        (ma for ma, diem in diem_theo_nguoi_dung.items() if diem >= diem_trung_binh),
        key=lambda ma: diem_theo_nguoi_dung[ma],
        reverse=True
    )

    print(f"\nNhung nguoi su dung co the ua thich bo phim {ma_so_phim} "
          f"(diem >= trung binh {diem_trung_binh:.2f}):")
    for ma in nguoi_dung_ua_thich:
        print(f"  Nguoi dung {ma} - Diem phu hop: {diem_theo_nguoi_dung[ma]:.2f}")
    print()
```
