```python
##------------------------------------------------------------------------------
## Viet doan chuong trinh cho phep nguoi su dung nhap danh sach sinh vien:
##    - ma so
##    - ho va chu lot
##    - ten
##    - gioi tinh: Nam, Nu
##    - ngay sinh
## va luu danh sach tren vao tap tin 'Sinh_vien.xls'
##
## Viet code:
import pandas as pd

danh_sach_sinh_vien = []
while True:
    ma_so = input("Nhap ma so sinh vien (Enter de dung): ")
    if ma_so == "":
        break
    ho_chu_lot = input("Nhap ho va chu lot: ")
    ten = input("Nhap ten: ")
    gioi_tinh = input("Nhap gioi tinh (Nam/Nu): ")
    ngay_sinh = input("Nhap ngay sinh (dd/mm/yyyy): ")
    danh_sach_sinh_vien.append({
        "Ma so": ma_so,
        "Ho va chu lot": ho_chu_lot,
        "Ten": ten,
        "Gioi tinh": gioi_tinh,
        "Ngay sinh": ngay_sinh
    })

folder1 = "C:/Users/DELL/Documents/Storage/UEH/Lập trình phân tích dữ liệu/[UEH] - DATA ANALYSIS CODING/Data/"
df_sinh_vien = pd.DataFrame(danh_sach_sinh_vien)
df_sinh_vien.to_excel(folder1 + "Sinh_vien.xlsx", index=False)
print(df_sinh_vien)
print("Da luu vao tap tin Sinh_vien.xlsx")
```

```python
##------------------------------------------------------------------------------
## Tập tin 'Titanic.xls' gồm có 2 sheets:
##   1) Passengers: lưu trữ thông tin của hành khách trên chuyến tàu Titanic
##                  với khóa nội và passengerID
##   2) Sinking   : lưu trữ số liệu thống kê về tai nạn chìm tàu Titanic
##      - survived: các giá trị 0 và 1 thể hiện lần lượt 'tử nạn' và 'sống sót'
##      - alive   : các giá trị 'no' và 'yes' thể hiện lần lượt 'đã chết' và
##                  'hiện còn sống' vào thời điểm thu thập số liệu
##      - alone   : các giá trị False và True thể hiện lần lượt 'có gia đình' và
##                  'độc thân' vào thời điểm thu thập số liệu
##------------------------------------------------------------------------------

## Luu y: doan code duoi day gia dinh tap tin 'Titanic.xls' da duoc tai len
## cung thu muc voi notebook nay.
TEN_TAP_TIN = "/Data/Titanic.xls"
folder = "C:/Users/DELL/Documents/Storage/UEH/Lập trình phân tích dữ liệu/[UEH] - DATA ANALYSIS CODING"
df_passengers = pd.read_excel(folder + TEN_TAP_TIN, sheet_name="Passengers")
df_sinking = pd.read_excel(folder + TEN_TAP_TIN, sheet_name="Sinking")
```

```python
## Liet ke cac columns co kieu NUMERICAL, CATEGORICAL cua sheet Passengers
##
## Viet code:
cot_numerical = df_passengers.select_dtypes(include=["number"]).columns.tolist()
cot_categorical = df_passengers.select_dtypes(include=["object", "category", "bool"]).columns.tolist()

print("Cac cot NUMERICAL:", cot_numerical)
print("Cac cot CATEGORICAL:", cot_categorical)
```

```python
## Liet ke danh sach nhung hanh khach la nu gioi va tren 60 tuoi
##
## Viet code:
nu_tren_60 = df_passengers[(df_passengers["sex"] == "female") & (df_passengers["age"] > 60)]
print(nu_tren_60)
```

```python
## Liet ke danh sach nhung hanh khach khong len tau o Southampton
##
## Viet code:
khong_o_southampton = df_passengers[df_passengers["embark_town"] != "Southampton"]
print(khong_o_southampton)
```

```python
## Cho biet so luong nu hanh khach hien con dang song va doc than
##
## Viet code:
df_hop_nhat = df_passengers.merge(df_sinking, on="passengerID", how="inner")
so_luong = len(df_hop_nhat[
    (df_hop_nhat["sex"] == "female") &
    (df_hop_nhat["alive"] == "yes") &
    (df_hop_nhat["alone"] == True)
])
print(f"So luong nu hanh khach con song va doc than: {so_luong}")
```

```python
## Cho biet hang ve co so tre em duoi 18 tuoi song sot cao nhat
##
## Viet code:
df_hop_nhat = df_passengers.merge(df_sinking, on="passengerID", how="inner")
tre_em_song_sot = df_hop_nhat[
    (df_hop_nhat["age"] < 18) & (df_hop_nhat["survived"] == 1)
]
dem_theo_hang = tre_em_song_sot.groupby("pclass").size()
hang_ve_cao_nhat = dem_theo_hang.idxmax()
print(f"So tre em song sot theo hang ve:\n{dem_theo_hang}")
print(f"Hang ve co so tre em song sot cao nhat: {hang_ve_cao_nhat}")
```
