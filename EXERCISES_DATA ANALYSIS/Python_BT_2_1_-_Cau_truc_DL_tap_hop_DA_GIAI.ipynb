#### **2. Cấu trúc STACK (LIFO)**
* Yêu cầu cài đặt bằng List
* Sử dụng thư viện collections.deque
---

```python
##------------------------------------------------------------------------------
## Cài đặt bằng list:
##     - mỗi phần tử chứa các thông tin (Id, f1, f2,..., fn)
##     - CUỐI list = ĐẦU stack (lật stack nằm ngang bên PHẢI)
##     - PUSH: thêm vào đầu stack --> append() vào cuối list
##     - POP : rút trích đầu stack --> pop() ở cuối list
##------------------------------------------------------------------------------
```

```python
##------------------------------------------------------------------------------
## Tao stack chua cac nodes cua do thi: node(Id, color, weight)
##------------------------------------------------------------------------------
## Viet code:
stack_nodes = []
stack_nodes.append([0, 'red', 12])
stack_nodes.append([1, 'green', -23.4])
stack_nodes.append([2, 'blue', 1.23])
stack_nodes.append([3, 'orange', 2.23])

print(f"Stack nodes: {stack_nodes}")
```

```python
##------------------------------------------------------------------------------
## Truy xuất (POP) tuần tự các phần tử trong stack_nodes
##------------------------------------------------------------------------------
stack_nodes = []
stack_nodes.append([0, 'red'   , 12])
stack_nodes.append([1, 'green' , -23.4])
stack_nodes.append([2, 'blue'  , 1.23])
stack_nodes.append([3, 'orange', 2.23])

## Viết code:


while stack_nodes:
    node = stack_nodes.pop()
    print(f"POP: {node}")
```

```python
##------------------------------------------------------------------------------
## Tạo stack chứa các edges của đồ thị: edge(Id, node_start, node_end, weight)
##------------------------------------------------------------------------------
stack_edges = []
stack_edges.append([0, 0, 1, 12])
stack_edges.append([1, 1, 2, -23.4])
stack_edges.append([2, 2, 1, 1.23])

## Viết code:


while stack_edges:
    edge = stack_edges.pop()
    print(f"POP: {edge}")
```

#### **3. Cấu trúc QUEUE (FIFO)**
* Yêu cầu cài đặt bằng List
* Có thể sử dụng thư viện collections.deque
---

```python
##------------------------------------------------------------------------------
## Cài đặt bằng list:
##     - mỗi phần tử chứa các thông tin (Id, f1, f2,..., fn)
##     - ENQUEUE: thêm vào cuối queue --> append() vào cuối list
##     - DEQUEUE: rút trích đầu stack --> pop() ở cuối list
##     (vào ở ĐUÔI, ra ở ĐẦU)
##------------------------------------------------------------------------------
```

```python
##------------------------------------------------------------------------------
## Truy xuất tuần tự các phần tử trong queue_nodes
##------------------------------------------------------------------------------
queue_nodes = []
queue_nodes.append([0, 'red'   , 12])
queue_nodes.append([1, 'green' , -23.4])
queue_nodes.append([2, 'blue'  , 1.23])
queue_nodes.append([3, 'orange', 2.23])

## Viết code:


while queue_nodes:
    node = queue_nodes.pop(0)   # rut trich o DAU list = DAU queue
    print(f"DEQUEUE: {node}")
```

```python
##------------------------------------------------------------------------------
## Tạo queue chứa các edges của đồ thị: edge(Id, node_start, node_end, weight)
##------------------------------------------------------------------------------
queue_edges = []
queue_edges.append([0, 0, 1, 12])
queue_edges.append([1, 1, 2, -23.4])
queue_edges.append([2, 2, 1, 1.23])

## Viết code:


while queue_edges:
    edge = queue_edges.pop(0)
    print(f"DEQUEUE: {edge}")
```

#### **Bai giai muc 4 - Cac bai tap trong giao trinh**
(Su dung danh sach mau: `a = [3, -5, 8, -2, 0, 7, -1, 4, 6, -9, 2]`)
---

```python
a = [3, -5, 8, -2, 0, 7, -1, 4, 6, -9, 2]
print("Danh sach mau:", a)
```

**1. Tong cac phan tu cua danh sach**

```python
tong = sum(a)
print(f"Tong = {tong}")
```

**2. Dem so luong so hang duong va tong cua chung**

```python
duong = [x for x in a if x > 0]
print(f"So luong so duong: {len(duong)}, Tong: {sum(duong)}")
```

**3. Trung binh cong ca danh sach va trung binh cong phan tu duong**

```python
tb_ca_ds = sum(a) / len(a)
duong = [x for x in a if x > 0]
tb_duong = sum(duong) / len(duong) if duong else 0
print(f"TB ca danh sach: {tb_ca_ds:.2f}, TB phan tu duong: {tb_duong:.2f}")
```

**4. Vi tri phan tu am dau tien**

```python
vi_tri = next((i for i, x in enumerate(a) if x < 0), -1)
print(f"Vi tri phan tu am dau tien: {vi_tri}")
```

**5. Vi tri phan tu duong cuoi cung**

```python
vi_tri = next((i for i in range(len(a) - 1, -1, -1) if a[i] > 0), -1)
print(f"Vi tri phan tu duong cuoi cung: {vi_tri}")
```

**6. Phan tu lon nhat va vi tri phan tu lon nhat cuoi cung**

```python
lon_nhat = max(a)
vi_tri_cuoi = len(a) - 1 - a[::-1].index(lon_nhat)
print(f"Lon nhat: {lon_nhat}, vi tri cuoi cung: {vi_tri_cuoi}")
```

**7. Phan tu lon thu nhi va cac vi tri dat gia tri do**

```python
gia_tri_khac_nhau = sorted(set(a), reverse=True)
if len(gia_tri_khac_nhau) >= 2:
    lon_thu_nhi = gia_tri_khac_nhau[1]
    vi_tri = [i for i, x in enumerate(a) if x == lon_thu_nhi]
    print(f"Lon thu nhi: {lon_thu_nhi}, cac vi tri: {vi_tri}")
else:
    print("Danh sach khong co du gia tri khac nhau")
```

**8. So luong cac so duong lien tiep nhieu nhat**

```python
max_dem, dem = 0, 0
for x in a:
    if x > 0:
        dem += 1
        max_dem = max(max_dem, dem)
    else:
        dem = 0
print(f"So luong so duong lien tiep nhieu nhat: {max_dem}")
```

**9. So luong cac so duong lien tiep co tong lon nhat**

```python
max_tong, tong_hien_tai, dem_hien_tai, max_dem = 0, 0, 0, 0
for x in a:
    if x > 0:
        tong_hien_tai += x
        dem_hien_tai += 1
        if tong_hien_tai > max_tong:
            max_tong = tong_hien_tai
            max_dem = dem_hien_tai
    else:
        tong_hien_tai, dem_hien_tai = 0, 0
print(f"Tong lon nhat: {max_tong}, so luong phan tu: {max_dem}")
```

**10. So luong cac phan tu lien tiep dan dau nhieu nhat**

```python
max_dem, dem = 1, 1
for i in range(1, len(a)):
    if a[i] * a[i - 1] < 0:
        dem += 1
        max_dem = max(max_dem, dem)
    else:
        dem = 1
print(f"So luong phan tu dan dau lien tiep nhieu nhat: {max_dem}")
```

**11. So luong cac phan tu khong tang nhieu nhat**

```python
max_dem, dem = 1, 1
for i in range(1, len(a)):
    if a[i] <= a[i - 1]:
        dem += 1
        max_dem = max(max_dem, dem)
    else:
        dem = 1
print(f"So luong phan tu khong tang lien tiep nhieu nhat: {max_dem}")
```

**12. Vi tri bat dau doan con duong lien tiep co nhieu phan tu nhat**

```python
max_dem, dem, vi_tri_bat_dau, vi_tri_tot_nhat = 0, 0, 0, 0
for i, x in enumerate(a):
    if x > 0:
        if dem == 0:
            vi_tri_bat_dau = i
        dem += 1
        if dem > max_dem:
            max_dem = dem
            vi_tri_tot_nhat = vi_tri_bat_dau
    else:
        dem = 0
print(f"Vi tri bat dau: {vi_tri_tot_nhat}, do dai: {max_dem}")
```

**13. Chuyen cac phan tu duong len dau danh sach**

```python
b = a.copy()
duong = [x for x in b if x > 0]
khong_duong = [x for x in b if x <= 0]
ket_qua = duong + khong_duong
print(f"Danh sach sau khi chuyen: {ket_qua}")
```

**14. So phan tu vua duong vua nguyen to va vi tri cua no**

```python
def la_so_nguyen_to(x):
    if x < 2:
        return False
    for i in range(2, int(x ** 0.5) + 1):
        if x % i == 0:
            return False
    return True

ket_qua = [(i, x) for i, x in enumerate(a) if x > 0 and la_so_nguyen_to(x)]
print(f"So luong: {len(ket_qua)}, (vi tri, gia tri): {ket_qua}")
```

**15. Chen so m vao dau, cuoi va vi tri thu 5 cua danh sach**

```python
b = a.copy()
m = 100
b.insert(0, m)      # chen vao dau
b.append(m)          # chen vao cuoi
b.insert(5, m)       # chen vao vi tri thu 5
print(f"Danh sach sau khi chen {m}: {b}")
```

**16. Chen danh sach [1,2,3] vao dau, cuoi va vi tri thu 5**

```python
b = a.copy()
chen = [1, 2, 3]
b = chen + b                 # chen vao dau
b = b + chen                 # chen vao cuoi
b[5:5] = chen                # chen vao vi tri thu 5
print(f"Danh sach sau khi chen: {b}")
```
