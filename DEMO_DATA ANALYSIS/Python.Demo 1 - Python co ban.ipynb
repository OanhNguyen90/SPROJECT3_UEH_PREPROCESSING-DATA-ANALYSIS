```python
## Nhập dữ liệu từ bàn phím: giá trị luôn có kiểu CHUỖI, dùng eval() để chuyển sang kiểu SỐ
r = eval(input('Cho biết bán kính r: '))
r
'Diện tích hình tròn có r =', r, 'là', r * r * 3.14
# Nếu không dùng hàm print() thì chỉ hiện kết quả của dòng lệnh cuối cùng
```

```python
## Sử dụng: print(f'{:}')
print(f'Diện tích hình tròn có bán kính r = {r:d} là {r * r * 3.14159}')
print(f'Diện tích hình tròn có bán kính r = {r:4d} là {r * r * 3.14159}')
print(f'Diện tích hình tròn có bán kính r = {r:d} là {r * r * 3.14159:.4f}')
print(f'Diện tích hình tròn có bán kính r = {r:d} là {r * r * 3.14159:.0f}')
```

```python
## Hiển thị các giá trị kiểu số
nb_float  = 2.123456789
nb_int    = 200
print(' a)', '%8d' %nb_int) ## độ dài 8
print(' b)', '{:4d}'.format(nb_int))
print('c1)', '%08d' %nb_int) ## thay space phía trước = 0
print('c2)', '{:08d}'.format(nb_int))
print(' d)', '%.4f' %nb_float) ## 4 chữ số thập phân
print(' e)', '{:f}'.format(nb_float)) ## ngầm định 6 chữ số thập phân
print('f1)', '{:10.4f}'.format(nb_float)) ## độ dài 10, 4 chữ số thập phân
print('f2)', '%*.*f' % (10, 4, nb_float)) ## độ dài 10, 4 chữ số thập phân
print('g1)', '%+d' %nb_int) ## có dấu +
print('g2)', '{:+d}'.format(nb_int)) ## có dấu +
print('h1)', '%d' % (-1 * nb_int)) ## có dấu -
print('h2)', '{:d}'.format(-1 * nb_int)) ## có dấu -
print(' i)', '{:=5d}'.format((-1 * nb_int))) ## thêm space giữa dấu và số
```

```python
## Hiển thị giá trị kiểu chuỗi
s1, s2 = 'Nước Việt Nam', 'Hà Nội'
### Canh TRÁI
print('a1)', '%-20s, %-20s' %(s1, s2)) ## độ dài 20
print('a2)', '{:20}, {:20}'.format(s1, s2)) ## độ dài 20
print(' b)', '{:_<20}, {:20}'.format(s1, s2)) ## thay space bằng _
### Canh PHẢI
print('c1)', '%20s, %20s' %(s1, s2)) ## độ dài 20
print('c2)', '{:>20}, {:>20}'.format(s1, s2)) ## độ dài 20
print(' d)', '{:_>20}, {:>20}'.format(s1, s2)) ## thay space bằng _
### Canh GIỮA
print(' e)', '{:^20}, {:^20}'.format(s1, s2)) ## độ dài 20
### Phối hợp TRÁI, PHẢI
print('f1)', '%-20s, %20s' %(s1, s2)) ## độ dài 20
print('f2)', '{:20}, {:>20}'.format(s1, s2)) ## độ dài 20
### Cắt chiều dài chuỗi
print('g1)', '%.8s' % (s1)) ## độ dài 8
print('g2)', '{:.8}'.format(s1)) ## độ dài 8
```

```python
## Hiển thị giá trị kiểu ngày
import datetime as dt
ngay   = dt.date(1984, 5, 17)
print('{:%Y/%m/%d %H:%M}'.format(ngay))
print('{:{dfmt} {tfmt}}'.format(ngay, dfmt='%d/%m/%Y', tfmt='%H:%M'))
print('{:{dfmt}} '.format(ngay, dfmt='%d/%m/%y'))
print('{:{dfmt}} '.format(ngay, dfmt='%d/%m/%Y'))
print('{:{dfmt}} '.format(ngay, dfmt='%d %b %Y'))
```

#### **3.1 Cấu trúc điều khiển IF**

```python
## Kiểm tra tính chẵn-lẻ của số nguyên
n = eval(input('Nhập một số nguyên: '))
if n % 2 == 0:
    print(str(n) + ' là số chẵn')
else:
    print(str(n) + ' là số lẻ')
```

```python
## Cấu trúc IF-ELIF-ELSE
so_bi_chia = eval(input('Nhập số bị chia: '))
so_chia    = eval(input('Nhập số chia: '))
if so_chia == 0:
    print('Không thể chia cho số 0')
elif (so_bi_chia % so_chia == 0):
    print(str(so_bi_chia) + ' chia hết cho ' + str(so_chia))
else:
    print(str(so_bi_chia) + ' không chia hết cho ' + str(so_chia))
```

```python
## Điều kiện rẽ nhánh bị trùng lắp
so_bi_chia = eval(input('Nhập số bị chia: '))
so_chia    = eval(input('Nhập số chia: '))
if so_chia == 0:
    print('Không thể chia cho số 0')
elif (so_bi_chia % so_chia == 0):
    print(str(so_bi_chia) + ' chia hết cho ' + str(so_chia))
elif (so_bi_chia == 10):
    print('Điều kiện bị bỏ qua')
else:
    print(str(so_bi_chia) + ' không chia hết cho ' + str(so_chia))
```

#### **3.2 Cấu trúc điều khiển MATCH-CASE**

```python
## Không có trường hợp DEFAULT
import datetime as dt
ngay = dt.date(2025, 8, 15)

match ngay.weekday():
    case 0:
        print('Monday')
    case 1:
        print('Tuesday')
    case 2:
        print('Wednesday')
    case 3:
        print('Thursday')
    case 4:
        print('Friday')
    case 5:
        print('Saturday')
    case 6:
        print('Sunday')
```

```python
## Có trường hợp DEFAULT
ngay = dt.date(2025, 8, 31)

match ngay.weekday():
    case 0:
        print('Monday')
    case 1:
        print('Tuesday')
    case 2:
        print('Wednesday')
    case 3:
        print('Thursday')
    case 4:
        print('Friday')
    case 5:
        print('Saturday')
    case _:
        print('Sunday')
```

```python
## Có thể sử dụng toán tử OR ( | )
match ngay.weekday():
    case 0 | 1 | 2 | 3 | 4:
        print(f'{ngay} là ngày LÀM VIỆC.')
    case 5 | 6:
        print(f'{ngay} là ngày NGHỈ.')
```

#### **3.3 Cấu trúc điều khiển FOR**

```python
## Dò tìm ký tự
chuoi = input('Nhập chuỗi: ')
ky_tu = input('Nhập ký tự cần tìm: ')
ky_tu = ky_tu[0:1] # chỉ lấy 1 ký tự

## Dò tìm ký tự trong chuỗi
tim_thay = False
for i in range (0, len(chuoi)):
    if (chuoi[i] == ky_tu):
        tim_thay = True ## đã tìm thấy
        break           ## không cần tiếp tục tìm
## Hiển thị kết quả
if tim_thay:
    print('Ký tự \'' + ky_tu + '\' CÓ trong chuỗi ' + '\'' + chuoi + '\'')
else:
    print('Ký tự \'' + ky_tu + '\' KHÔNG có trong chuỗi ' + '\'' + chuoi + '\'')
```

#### **4.1 Thủ tục (procedure)**

#### **4.2 Hàm (function)**

```python
## Hàm tính giai thừa bằng vòng lặp
def giai_thua_for(n):
    giai_thua = 1
    for i in range(1, n + 1):
        giai_thua *= i

    return giai_thua
```

```python
## Gọi hàm
n = int(input('Nhập số nguyên (n < 10) '))
print('Sử dụng vòng lặp:', str(n) + '! =', giai_thua_for(n))
```

```python
## Hàm có thể gọi thực hiện chính nó: ĐỆ QUY
def giai_thua_de_quy(n):
    if(n <= 1):
        return 1

    return(n * giai_thua_de_quy(n - 1))
```

```python
## Gọi hàm
print('Sử dụng đệ quy :', str(n) + '! =', giai_thua_de_quy(n))
```

```python
## Hàm có thể trả về NHIỀU kết quả
def hinh_tron(ban_kinh):
    chu_vi    = (ban_kinh * 2) * 3.14
    tich_so = (ban_kinh ** 2) * 3.14

    return chu_vi, tich_so
```

```python
## Gọi hàm
r      = eval(input('Cho biết bán kính '))
cv, dt = hinh_tron(r)
print(f'Chu vi hình tròn = {cv:.2f}, diện tích hình tròn = {dt:.2f}')
```

```python
## Kết quả có thể có kiểu dữ liệu tùy ý
def tieu_de(s, w):
    return [s.ljust(w), s.center(w)], s.rjust(w) ## canh biên: trái, giữa, phải
```

```python
x, y = tieu_de('Cộng Hòa Xã Hội Chủ Nghĩa Việt Nam', 50) ## Canh biên tiêu đề với độ rộng 50
print('Tiêu đề trái: ' + x[0])
print('Tiêu đề phải: ' + x[1])
print('Tiêu đề giữa: ' + y)
```

```python
## Giá trị ngầm định cho tham số
def hinh_tron(ban_kinh, PI = 3.14):
    chu_vi    = (ban_kinh * 2) * PI
    tich_so = (ban_kinh ** 2) * PI

    return chu_vi, tich_so
```

```python
## Gọi hàm với ĐẦY ĐỦ giá trị cho tham số
cv, dt = hinh_tron(r, 3.1416)
print(f'Chu vi hình tròn = {cv:.4f}, diện tích hình tròn = {dt:.4f}')

## Gọi hàm với giá trị tham số NGẦM ĐỊNH
cv, dt = hinh_tron(r)
print(f'Chu vi hình tròn = {cv:.4f}, diện tích hình tròn = {dt:.4f}')
```

```python
## Gọi hàm với THỨ TỰ LINH HOẠT của các tham số
cv, dt = hinh_tron(PI = 3.14, ban_kinh = 7.8)
print(f'Chu vi hình tròn = {cv:.2f}, diện tích hình tròn = {dt:.2f}')
```

```python
## Truyền tham số theo GIÁ TRỊ (by VALUE)
def by_value(param1, param2):
    param1 *= 2
    result  = param1 * param2

    return result
```

```python
## Gọi thực hiện
x, y  = 2, 8
value = by_value(x, y)
print(f'x = {x}')
```

```python
## Truyền tham số theo ĐỐI TƯỢNG THAM CHIẾU (by OBJECT REFERENCE)
def by_reference(param1, param2):
    param1[0] = 27
    result    = param1 * param2

    return result
```

```python
## Gọi thực hiện
lst, y  = [1, 2, 3], 8
value = by_reference(lst, y)
print(f'List = {lst}')
```

```python
## Sử dụng *args (to pass a non-keyworded, variable-length argument list)
##    với SỐ LƯỢNG LINH HOẠT các tham số
def func1(*args):
    for arg in args:
        print (arg)
```

```python
## Gọi thực hiện với tham số *argv
func1('Python', 'Programming', 'Language')
```

```python
func1('Python', 'Language')
```

```python
## Sử dụng **kwargs (to pass a keyworded, variable-length argument list)
def func2(**kwargs):
    for kw, value in kwargs.items():
        print (f'{kw} = {value}')
```

```python
func2(prm1 = 'Python', prm2 = 'Programming', prm3 = 'Language')
```

#### **4.3 Hàm lambda (anonymous function)**
* phương pháp lập trình hàm (<i>functional programming</i>)
* dạng công thức được lưu lại (tái sử dụng)

---

```python
## Hàm lambda đơn giản
tich_so = lambda a, b: (a * b)  # có thể xem 'tich_so' là tên hàm

print(f'Tích số của 9 và 2 = {tich_so(9, 2)}')
print(f'Tích số của 8 và 8 = {tich_so(8, 8)}')
```

```python
## Sử dụng giá trị ngầm định
dien_tich_hinh_tron = lambda r, PI = 3.14: (r ** 2) * PI

print(f'Diện tích hình tròn có bán kính 10 = {dien_tich_hinh_tron(10, 3.1416):.4f}')
print(f'Diện tích hình tròn có bán kính 10 = {dien_tich_hinh_tron(10):.4f}')
```

```python
## Hàm lambda với điều kiện if-else
nb = lambda x: 'số CHẴN' if (x % 2) == 0 else 'số LẺ'

print(f'100 là {nb(100)}')
print(f'203 là {nb(203)}')
```
