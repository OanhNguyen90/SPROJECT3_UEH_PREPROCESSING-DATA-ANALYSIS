### **2. Regular Expression (RegEx)**
---

```python
## https://docs.python.org/3/library/re.html
## https://www.w3schools.com/python/python_regex.asp
import re

## Hàm re.findall():
##   - trả về những dãy ký tự (list) phù hợp với mẫu tìm kiếm (pattern),
##     và liệt kê theo thứ tự xuất hiện của các dãy ký tự
##   - các dãy ký tự có thể xuất hiện nhiều lần

## Hàm re.search():
##   - trả về đối tượng có kiểu match nếu tìm thấy khớp với pattern
##     ở vị trí TÙY Ý trong chuỗi

## Hàm re.match():
##   - trả về đối tượng có kiểu match nếu tìm thấy khớp với pattern
##     ở ví trí BẮT ĐẦU của chuỗi
```

```python
txt1   = 'Python 3.13 was released on October 7, 2024.'
txt2   = 'Hanoi, Hue, HoChiMinh City, Vietnam'
txt3   = 'BaaaA CaaA'
n1, n2 = 10, 20 # số lượng kết quả tối đa
```

#### **2.1 Metacharacters**

##### <b>[ ]</b>: <i>Sử dụng pattern chứa một tập hợp (SET) các KÝ TỰ</i>




    




```python
## VD: pattern đơn giản gồm danh sách các ký tự (không liên tục)
result = re.findall('[aze]', txt1) # những ký tự a, z, e
if result: # type(result) = list; điều kiện đúng --> (!= [])
    print(f'Tìm thấy những ký tự a, z, e: {result[:n1]}')
else:
    print('Chuỗi không khớp với mẫu tìm kiếm !')
```

```python
## Phủ định
result = re.findall('[^aze]', txt1) # những ký tự không phải a, z, e
if result:
    print(f'Tìm thấy những ký tự không phải a, z, e: {result[:n1]}')
else:
    print('Chuỗi không khớp với mẫu tìm kiếm !')
```

```python
## VD: pattern đơn giản gồm danh sách các ký tự (liên tục)
result = re.findall('[a-n]', txt2) # những ký tự từ a đến n (a > n)
if result:
    print(f'Tìm thấy những ký tự từ a đến n: {result[:n1]}')
else:
    print('Chuỗi không khớp với mẫu tìm kiếm !')
```

```python
## Phủ định
result = re.findall('[^a-n]', txt1) # những ký tự không phải từ a đến n
if result:
    print(f'Tìm thấy những ký tự không phải từ a đến n: {result[:n1]}')
else:
    print('Chuỗi không khớp với mẫu tìm kiếm !')
```

```python
## VD: pattern kết hợp nhiều dãy
result = re.findall('[a-n,M-Z]', txt1) # ký tự a-n, M-Z
if result:
    print(f'Tìm thấy ký tự a-n hay M-Z: {result[:n1]}')
else:
    print('Chuỗi không khớp với mẫu tìm kiếm !')
```

```python
## Có thể bỏ dấu , trong pattern
result = re.findall('[a-nM-Z]', txt1) # ký tự a-n, M-Z
if result:
    print(f'Tìm thấy ký tự a-n hay M-Z: {result[:n1]}')
else:
    print('Chuỗi không khớp với mẫu tìm kiếm !')
```

```python
## Áp dụng trên chữ số
result = re.findall('[0-2][1-4]', txt1) # số gồm 2 chữ số từ 0 đến 2 và từ 2 đến 4
if result:
    print(f'Tìm thấy số gồm 2 chữ số từ 0 đến 2 và từ 2 đến 4: {result[:n1]}')
else:
    print('Chuỗi không khớp với mẫu tìm kiếm !')
```

##### <b>.</b>: <i>Đại diện cho 1 ký tự bất kỳ</i>

```python
## VD: mỗi dấu . trong pattern đại diện cho 1 ký tự bất kỳ
result = re.findall('.o..', txt1)
if result:
    print(f'Kết quả: {result[:n1]}')
else:
    print('Chuỗi không khớp với mẫu tìm kiếm !')
```

```python
## Tìm không thấy
result = re.findall('io..', txt1)
if result:
    print(f'Kết quả: {result[:n1]}')
else:
    print('Chuỗi không khớp với mẫu tìm kiếm !')
```

##### <b>*</b> (quantifier): <i>Số lần ký tự đứng trước được lặp lại tùy ý hoặc không xuất hiện</i>

```python
txt1   = 'Python 3_13 was released on October 7, 2024.'
txt2   = 'Hanoi, Hue, HoChiMinh City, Vietnam'
txt3   = 'BaaaA CaaA'
n1, n2 = 10, 10 # số lượng kết quả tối đa
```

```python
## VD: Áp dụng với 1 ký tự cụ thể là 'n'
result = re.findall('.on*', txt1) # 'n' có thể lặp lại tùy ý hoặc không xuất hiện
if result:
    print(f'Kết quả: {result[:n1]}')
else:
    print('Chuỗi không khớp với mẫu tìm kiếm !')
```

```python
## VD: Áp dụng với 1 ký tự bất kỳ (.)
result = re.findall('o.*e', txt1) # theo sau 'o' là dãy ký tự tùy ý rồi đến 'e'
                                  # vì '*' là tùy ý nên kéo dài đến 'e' xa nhất !

if result:
    print(f'Kết quả: {result[:n1]}')
else:
    print('Chuỗi không khớp với mẫu tìm kiếm !')
```

##### <b>+</b> (quantifier): <i>Số lần ký tự đứng trước được lặp lại tùy ý với ít nhất 1 lần

```python
## VD: Áp dụng với 1 ký tự cụ thể là 'n'
result = re.findall('.on+', txt1) # 'n' xuất hiện ít nhất 1 lần
if result:
    print(f'Kết quả: {result[:n1]}')
else:
    print('Chuỗi không khớp với mẫu tìm kiếm !')
```

##### <b>?</b> (quantifier): <i>Ký tự đứng trước xuất hiện 1 lần hoặc không xuất hiện</i>

```python
## VD: Áp dụng với 1 ký tự cụ thể là 'n'
result = re.findall('.on?', txt1) # 'n' có thể xuất hiện 1 lần hoặc không xuất hiện
if result:
    print(f'Kết quả: {result[:n1]}')
else:
    print('Chuỗi không khớp với mẫu tìm kiếm !')
```

##### <b>{ }</b> (quantifier): <i>Chỉ định số lần ký tự đứng trước được xuất hiện</i>

```python
## VD: Áp dụng với 1 ký tự cụ thể, với số lần xuất hiện > 0
result = re.findall('a{3}.', txt3) # 'a' xuất hiện 3 lần, theo sau là 1 ký tự tùy ý
if result:
    print(f'Kết quả: {result[:n1]}')
else:
    print('Chuỗi không khớp với mẫu tìm kiếm !')
```

```python
## VD: Áp dụng với 1 ký tự cụ thể, với số lần xuất hiện = 0
result = re.findall('a{0}C', txt3) # 'a' không xuất hiện trước 'C'
if result:
    print(f'Kết quả: {result[:n1]}')
else:
    print('Chuỗi không khớp với mẫu tìm kiếm !')
```

```python
## VD: Áp dụng với 1 ký tự cụ thể, với số lần xuất hiện < 0
result = re.findall('a{-1}', txt3)
if result:
    print(f'Kết quả: {result[:n1]}')
else:
    print('Chuỗi không khớp với mẫu tìm kiếm !')
```

```python
## VD: Áp dụng với 1 ký tự bất kỳ (.), với số lần xuất hiện > 0
result = re.findall('.{3}o', txt1) # phía trước 'o' có 3 ký tự
if result:
    print(f'Kết quả: {result[:n1]}')
else:
    print('Chuỗi không khớp với mẫu tìm kiếm !')
```

#### <b>2.2 Special sequences (\\)</b>

```python
##------------------------------------------------------------------------------
## RAW string (r'...'):
##------------------------------------------------------------------------------
## Chú ý sử dụng khi trong pattern có \
## When an 'r' or 'R' prefix is present, a character following a backslash is included
## in the string without change, and all backslashes are left in the string.
```

##### <b>\w</b>: <i>Đại diện cho ký tự (character), chữ số (digit) hay dấu _ (underscore)</i>

```python
## VD:
result = re.findall(r'\w', txt1) # những ký tự, chữ số hay _
if result:
    print(f'Kết quả: {result}') # không có dấu , và dấu .
else:
    print('Chuỗi không khớp với mẫu tìm kiếm !')
```

##### <b>\W</b>: <i>Ngược lại với \w</i>

```python
## VD:
result = re.findall(r'\W', txt1) # không phải là ký tự, chữ số hay _
if result:
    print(f'Kết quả: {result}') # dấu , và dấu và khoảng trắng
else:
    print('Chuỗi không khớp với mẫu tìm kiếm !')
```

##### <b>\A</b>: <i>So khớp STRING được BẮT ĐẦU bằng dãy ký tự</i>

```python
## So sánh với ^ (metacharacter)
result = re.search('^Ha', txt2) # chuỗi bắt dầu bằng Ha
if result:
    print(f'Chuỗi \'{result.string}\' bắt đầu bằng Ha')
else:
    print('Chuỗi không khớp với mẫu tìm kiếm !')
```

```python
## VD:
result = re.findall(r'Vietnam\Z', txt2)
if result:
    print(f'Chuỗi \'{txt2}\' kết thúc bằng \'{result}\'')
else:
    print('Chuỗi không khớp với mẫu tìm kiếm !')
```
